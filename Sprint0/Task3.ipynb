<h3 style="color:DodgerBlue">Описание задачи:</h3>
<p>Создать базовый класс <strong>Ticket</strong> в C#, который будет представлять билеты на различные мероприятия. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.
</p>

<h3 style="color:DodgerBlue">Требования к базовому классу Ticket</h3>
<ul>
    <li><strong>Атрибуты:</strong> ID билета (TicketId), Номер ряда (RowNumber), Номер места (SeatNumber).
</li>
    <li><strong>Методы:</strong>
        <ul>
            <li><code>PrintTicketInfo()</code>: метод для печати информации о билете.</li>
            <li><code>ReserveSeat()</code>: метод для резервирования места.</li>
            <li><code>GetTicketDetails()</code>: метод для получения деталей билета.</li>
        </ul>
    </li>
</ul>

<h3 style="color:DodgerBlue">Требования к производным классаам:</h3>
<ol>
    <li><strong>БилетНаКонцерт (ConcertTicket):</strong>
        <ul>
            <li>Должен содержать дополнительные атрибуты, такие как Имя исполнителя (ArtistName).</li>
            <li>Метод <code>PrintTicketInfo()</code> должен быть переопределен для включения имени исполнителя в информацию о билете.</li>
        </ul>
    </li>
    <li><strong>БилетНаСпектакль (PlayTicket):</strong>
        <ul>
            <li>Должен содержать дополнительные атрибуты, такие как Название спектакля (PlayTitle).</li>
            <li>Метод <code>GetTicketDetails()</code> должен быть переопределен для отображения названия спектакля вместе с другими деталями билета.</li>
        </ul>
    </li>
    <li><strong>БилетНаКонференцию (ConferenceTicket) (если требуется третий класс):</strong>
        <ul>
            <li>Должен содержать дополнительные атрибуты, такие как Название конференции (ConferenceName).</li>
            <li>Метод <code>ReserveSeat()</code> должен быть переопределен для добавления информации о предпочтениях участника относительно мест.</li>
        </ul>
    </li>
</ol>
