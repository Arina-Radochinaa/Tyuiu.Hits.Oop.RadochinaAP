### Структура и Объявление Класса в C#

В C#, класс объявляется с использованием ключевого слова `class`, за которым следует имя класса. Класс может содержать поля (данные), свойства (способ доступа к данным), методы (функциональность) и события. 

#### Свойства

Свойства в C# используются для инкапсуляции данных. Они предоставляют гибкий механизм для чтения, записи или вычисления значений приватных полей. Свойства могут иметь два акцессора: `get` (геттер) для чтения значения и `set` (сеттер) для его изменения.

#### Геттеры и Сеттеры

Геттеры и сеттеры являются частью свойств класса в C#. Геттер (`get`) используется для возврата значения свойства, а сеттер (`set`) — для установки нового значения. Сеттер может включать логику валидации или другие операции перед изменением значения поля.

### Пример Классов Vehicle и Car

Давайте рассмотрим примеры классов `Vehicle` и `Car`, чтобы проиллюстрировать использование свойств, геттеров и сеттеров, а также разницу между статическими и обычными атрибутами и методами.

#### Класс Vehicle

```csharp
public class Vehicle
{
    private string _color;
    private int _speed;

    // Свойство Color с геттером и сеттером
    public string Color
    {
        get { return _color; }
        set { _color = value; }
    }

    // Свойство Speed с геттером и сеттером
    public int Speed
    {
        get { return _speed; }
        set
        {
            if (value >= 0)
                _speed = value;
            else
                throw new ArgumentOutOfRangeException("Скорость не может быть отрицательной!");
        }
    }

    // Статический атрибут
    public static int TotalVehicles { get; private set; } = 0;

    // Конструктор
    public Vehicle()
    {
        TotalVehicles++;
    }

    // Обычный метод
    public void Accelerate(int delta)
    {
        Speed += delta;
    }

    // Статический метод
    public static void DisplayTotalVehicles()
    {
        Console.WriteLine($"Общее количество транспортных средств: {TotalVehicles}");
    }
}
```

#### Класс Car

```csharp
public class Car : Vehicle
{
    private int _fuelLevel;

    // Свойство FuelLevel с геттером и сеттером
    public int FuelLevel
    {
        get { return _fuelLevel; }
        set { _fuelLevel = value; }
    }

    // Конструктор
    public Car() : base()
    {
        FuelLevel = 100; // Задаем начальный уровень топлива
    }

    // Обычный метод
    public void Refuel(int amount)
    {
        FuelLevel += amount;
    }
}
```

### Разница между Статическими и Обычными Атрибутами и Методами Класса

- **Статические атрибуты и методы** принадлежат самому классу, а не его экземплярам. Они могут быть вызваны без создания объекта класса. В примере выше `TotalVehicles` является статическим атрибутом, а `DisplayTotalVehicles()` — статическим методом класса `Vehicle`.
- **Обычные атрибуты и методы** принадлежат экземплярам класса и могут иметь разные значения для разных объектов. В примере выше `_color`, `_speed` и `_fuelLevel` являются обычными атрибутами, а методы `Accelerate()` и `Refuel()` — обычными методами.

````
using System;

public class Animal
{
    private string _name;
    private string _species;
    private int _age;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public string Species
    {
        get { return _species; }
        set { _species = value; }
    }

    public int Age
    {
        get { return _age; }
        set
        {
            if (value >= 0)
                _age = value;
            else
                throw new ArgumentOutOfRangeException("Возраст не может быть отрицательным!");
        }
    }

    public static int TotalAnimals { get; private set; } = 0;

    public Animal()
    {
        _name = "Безымянный";
        _species = "Неизвестно";
        _age = 0;
        TotalAnimals++;
    }

    public Animal(string name, string species, int age)
    {
        _name = name;
        _species = species;
        _age = age;
        TotalAnimals++;
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"{_name} издает звук...");
    }

    public void Eat()
    {
        Console.WriteLine($"{_name} кушает...");
    }

    public void Sleep()
    {
        Console.WriteLine($"{_name} спит...");
    }

    public static void DisplayTotalAnimals()
    {
        Console.WriteLine($"Общее количество животных: {TotalAnimals}");
    }
}

public class Dog : Animal
{
    private string _breed;
    private int _energyLevel;

    public string Breed
    {
        get { return _breed; }
        set { _breed = value; }
    }

    public int EnergyLevel
    {
        get { return _energyLevel; }
        set
        {
            if (value >= 0 && value <= 100)
                _energyLevel = value;
            else
                throw new ArgumentOutOfRangeException("Уровень энергии должен быть от 0 до 100!");
        }
    }

    public Dog() : base()
    {
        _breed = "Дворняга";
        _energyLevel = 50;
    }

    public Dog(string name, string species, int age, string breed, int energyLevel) 
        : base(name, species, age)
    {
        _breed = breed;
        EnergyLevel = energyLevel;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Гав-гав!");
    }

    public void Fetch()
    {
        Console.WriteLine($"{Name} приносит палку!");
        _energyLevel -= 10;
        Console.WriteLine($"Уровень энергии: {_energyLevel}");
    }
}

public class Cat : Animal
{
    private bool _isIndoor;
    private int _livesLeft;

    public bool IsIndoor
    {
        get { return _isIndoor; }
        set { _isIndoor = value; }
    }

    public int LivesLeft
    {
        get { return _livesLeft; }
        set
        {
            if (value >= 0 && value <= 9)
                _livesLeft = value;
            else
                throw new ArgumentOutOfRangeException("Количество жизней должно быть от 0 до 9!");
        }
    }

    public Cat() : base()
    {
        _isIndoor = true;
        _livesLeft = 9;
    }

    public Cat(string name, string species, int age, bool isIndoor, int livesLeft) 
        : base(name, species, age)
    {
        _isIndoor = isIndoor;
        LivesLeft = livesLeft;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Мяу-мяу!");
    }

    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет...");
    }
}

public class Bird : Animal
{
    private double _wingspan;
    private bool _canFly;

    public double Wingspan
    {
        get { return _wingspan; }
        set { _wingspan = value; }
    }

    public bool CanFly
    {
        get { return _canFly; }
        set { _canFly = value; }
    }

    public Bird() : base()
    {
        _wingspan = 0;
        _canFly = true;
    }

    public Bird(string name, string species, int age, double wingspan, bool canFly) 
        : base(name, species, age)
    {
        _wingspan = wingspan;
        _canFly = canFly;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} поет: Чик-чирик!");
    }

    public void Fly()
    {
        if (_canFly)
            Console.WriteLine($"{Name} летает в небе!");
        else
            Console.WriteLine($"{Name} не умеет летать!");
    }
}

Animal defaultAnimal = new Animal();
Console.WriteLine("=== ЖИВОТНОЕ ПО УМОЛЧАНИЮ ===");
Console.WriteLine($"Имя: {defaultAnimal.Name}, Вид: {defaultAnimal.Species}, Возраст: {defaultAnimal.Age}");
defaultAnimal.MakeSound();
defaultAnimal.Eat();
defaultAnimal.Sleep();
Console.WriteLine();

Dog dog = new Dog("Бобик", "Собака", 5, "Лабрадор", 80);
Console.WriteLine("=== СОБАКА ===");
Console.WriteLine($"Имя: {dog.Name}, Вид: {dog.Species}, Возраст: {dog.Age}, Порода: {dog.Breed}, Энергия: {dog.EnergyLevel}");
dog.MakeSound();
dog.Eat();
dog.Sleep();
dog.Fetch();
Console.WriteLine();

Cat cat = new Cat("Мурка", "Кошка", 3, true, 9);
Console.WriteLine("=== КОШКА ===");
Console.WriteLine($"Имя: {cat.Name}, Вид: {cat.Species}, Возраст: {cat.Age}, Домашняя: {cat.IsIndoor}, Жизней: {cat.LivesLeft}");
cat.MakeSound();
cat.Eat();
cat.Sleep();
cat.Purr();
Console.WriteLine();

Bird bird = new Bird("Кеша", "Попугай", 2, 25.5, true);
Console.WriteLine("=== ПТИЦА ===");
Console.WriteLine($"Имя: {bird.Name}, Вид: {bird.Species}, Возраст: {bird.Age}, Размах крыльев: {bird.Wingspan} см, Может летать: {bird.CanFly}");
bird.MakeSound();
bird.Eat();
bird.Sleep();
bird.Fly();
Console.WriteLine();

Animal.DisplayTotalAnimals();
````
