```
public class Ticket
{
    private int _ticketId;
    private int _rowNumber;
    private int _seatNumber;
    private bool _isReserved;

    public int TicketId
    {
        get { return _ticketId; }
        set { _ticketId = value; }
    }

    public int RowNumber
    {
        get { return _rowNumber; }
        set
        {
            if (value > 0)
                _rowNumber = value;
            else
                throw new ArgumentOutOfRangeException("Номер ряда должен быть больше 0!");
        }
    }

    public int SeatNumber
    {
        get { return _seatNumber; }
        set
        {
            if (value > 0)
                _seatNumber = value;
            else
                throw new ArgumentOutOfRangeException("Номер места должен быть больше 0!");
        }
    }

    public bool IsReserved
    {
        get { return _isReserved; }
        protected set { _isReserved = value; }
    }

    public static int TotalTickets { get; private set; } = 0;

    public Ticket(int ticketId, int rowNumber, int seatNumber)
    {
        _ticketId = ticketId;
        _rowNumber = rowNumber;
        _seatNumber = seatNumber;
        _isReserved = false;
        TotalTickets++;
    }

    public virtual void PrintTicketInfo()
    {
        Console.WriteLine($"Билет №{_ticketId}");
        Console.WriteLine($"Ряд: {_rowNumber}, Место: {_seatNumber}");
        Console.WriteLine($"Статус: {(_isReserved ? "Зарезервирован" : "Свободен")}");
    }

    public virtual void ReserveSeat()
    {
        if (!_isReserved)
        {
            _isReserved = true;
            Console.WriteLine($"Место {_seatNumber} в ряду {_rowNumber} успешно зарезервировано.");
        }
        else
        {
            Console.WriteLine($"Место {_seatNumber} в ряду {_rowNumber} уже зарезервировано.");
        }
    }

    public virtual string GetTicketDetails()
    {
        return $"ID: {_ticketId}, Ряд: {_rowNumber}, Место: {_seatNumber}, Статус: {(_isReserved ? "Зарезервирован" : "Свободен")}";
    }

    public static void DisplayTotalTickets()
    {
        Console.WriteLine($"Общее количество билетов: {TotalTickets}");
    }
}

public class ConcertTicket : Ticket
{
    private string _artistName;

    public string ArtistName
    {
        get { return _artistName; }
        set { _artistName = value; }
    }

    public ConcertTicket(int ticketId, int rowNumber, int seatNumber, string artistName)
        : base(ticketId, rowNumber, seatNumber)
    {
        _artistName = artistName;
    }

    public override void PrintTicketInfo()
    {
        base.PrintTicketInfo();
        Console.WriteLine($"Исполнитель: {_artistName}");
    }

    public override string GetTicketDetails()
    {
        return $"{base.GetTicketDetails()}, Исполнитель: {_artistName}";
    }

    public void MeetAndGreet()
    {
        Console.WriteLine($"У вас есть возможность встретиться с {_artistName} после концерта!");
    }
}

public class PlayTicket : Ticket
{
    private string _playTitle;

    public string PlayTitle
    {
        get { return _playTitle; }
        set { _playTitle = value; }
    }

    public PlayTicket(int ticketId, int rowNumber, int seatNumber, string playTitle)
        : base(ticketId, rowNumber, seatNumber)
    {
        _playTitle = playTitle;
    }

    public override string GetTicketDetails()
    {
        return $"{base.GetTicketDetails()}, Спектакль: {_playTitle}";
    }

    public void ShowProgram()
    {
        Console.WriteLine($"Программа спектакля «{_playTitle}»:");
        Console.WriteLine("Акт 1, Антракт, Акт 2");
    }
}

public class ConferenceTicket : Ticket
{
    private string _conferenceName;
    private string _seatPreference;

    public string ConferenceName
    {
        get { return _conferenceName; }
        set { _conferenceName = value; }
    }

    public string SeatPreference
    {
        get { return _seatPreference; }
        set { _seatPreference = value; }
    }

    public ConferenceTicket(int ticketId, int rowNumber, int seatNumber,
                            string conferenceName, string seatPreference)
        : base(ticketId, rowNumber, seatNumber)
    {
        _conferenceName = conferenceName;
        _seatPreference = seatPreference;
    }

    public override void ReserveSeat()
    {
        base.ReserveSeat();
        Console.WriteLine($"Предпочтения участника: {_seatPreference}");
    }

    public override string GetTicketDetails()
    {
        return $"{base.GetTicketDetails()}, Конференция: {_conferenceName}, Предпочтения: {_seatPreference}";
    }
}

// ============ ПРОСТОЕ НАСЛЕДОВАНИЕ ============
public class VipConcertTicket : ConcertTicket
{
    private string _loungeName;

    public string LoungeName
    {
        get { return _loungeName; }
        set { _loungeName = value; }
    }

    public VipConcertTicket(int ticketId, int rowNumber, int seatNumber,
                            string artistName, string loungeName)
        : base(ticketId, rowNumber, seatNumber, artistName)
    {
        _loungeName = loungeName;
    }

    public override void PrintTicketInfo()
    {
        base.PrintTicketInfo();
        Console.WriteLine($"VIP-лаунж: {_loungeName}");
    }

    public void AccessLounge()
    {
        Console.WriteLine($"Доступ в VIP-лаунж «{_loungeName}» разрешён!");
    }
}

// ============ СЛОЖНОЕ НАСЛЕДОВАНИЕ ============
public class BusinessConferenceTicket : ConferenceTicket
{
    private bool _hasPrivateMeeting;

    public bool HasPrivateMeeting
    {
        get { return _hasPrivateMeeting; }
        set { _hasPrivateMeeting = value; }
    }

    public BusinessConferenceTicket(int ticketId, int rowNumber, int seatNumber,
                                    string conferenceName, string seatPreference,
                                    bool hasPrivateMeeting)
        : base(ticketId, rowNumber, seatNumber, conferenceName, seatPreference)
    {
        _hasPrivateMeeting = hasPrivateMeeting;
    }

    public override void PrintTicketInfo()
    {
        base.PrintTicketInfo();
        Console.WriteLine($"Личная встреча: {_hasPrivateMeeting}");
    }

    public void ScheduleMeeting()
    {
        Console.WriteLine($"Запланирована личная встреча: {_hasPrivateMeeting}");
    }
}

// ============ МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ (через интерфейсы) ============
public interface IDiscountable
{
    decimal Discount { get; set; }
    decimal GetDiscountedPrice(decimal basePrice);
}

public interface ITransferable
{
    string TransferTo { get; set; }
    void Transfer();
}

public class FlexibleTicket : Ticket, IDiscountable, ITransferable
{
    public decimal Discount { get; set; }
    public string TransferTo { get; set; }

    public FlexibleTicket(int ticketId, int rowNumber, int seatNumber,
                          decimal discount, string transferTo)
        : base(ticketId, rowNumber, seatNumber)
    {
        Discount = discount;
        TransferTo = transferTo;
    }

    public decimal GetDiscountedPrice(decimal basePrice)
    {
        return basePrice - (basePrice * Discount / 100);
    }

    public void Transfer()
    {
        Console.WriteLine($"Билет №{TicketId} передан: {TransferTo}");
    }
}

ConcertTicket concert = new ConcertTicket(1, 5, 12, "Imagine Dragons");
PlayTicket play = new PlayTicket(2, 3, 7, "Гамлет");
ConferenceTicket conference = new ConferenceTicket(3, 10, 25, "DevOps Conf 2026", "У окна");
VipConcertTicket vipConcert = new VipConcertTicket(4, 1, 1, "Metallica", "Golden Lounge");
BusinessConferenceTicket businessConf = new BusinessConferenceTicket(5, 8, 15, "Tech Summit", "У прохода", true);
FlexibleTicket flexible = new FlexibleTicket(6, 4, 9, 15, "Иван Петров");

Console.WriteLine("=== БИЛЕТ НА КОНЦЕРТ ===");
concert.PrintTicketInfo();
concert.ReserveSeat();
Console.WriteLine(concert.GetTicketDetails());
concert.MeetAndGreet();
Console.WriteLine();

Console.WriteLine("=== БИЛЕТ НА СПЕКТАКЛЬ ===");
play.PrintTicketInfo();
play.ReserveSeat();
Console.WriteLine(play.GetTicketDetails());
play.ShowProgram();
Console.WriteLine();

Console.WriteLine("=== БИЛЕТ НА КОНФЕРЕНЦИЮ ===");
conference.PrintTicketInfo();
conference.ReserveSeat();
Console.WriteLine(conference.GetTicketDetails());
Console.WriteLine();

Console.WriteLine("=== ПРОСТОЕ НАСЛЕДОВАНИЕ: VIP-КОНЦЕРТ ===");
vipConcert.PrintTicketInfo();
vipConcert.ReserveSeat();
Console.WriteLine(vipConcert.GetTicketDetails());
vipConcert.MeetAndGreet();
vipConcert.AccessLounge();
Console.WriteLine();

Console.WriteLine("=== СЛОЖНОЕ НАСЛЕДОВАНИЕ: БИЗНЕС-КОНФЕРЕНЦИЯ ===");
businessConf.PrintTicketInfo();
businessConf.ReserveSeat();
Console.WriteLine(businessConf.GetTicketDetails());
businessConf.ScheduleMeeting();
Console.WriteLine();

Console.WriteLine("=== МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ: ГИБКИЙ БИЛЕТ ===");
flexible.PrintTicketInfo();
Console.WriteLine($"Цена со скидкой: {flexible.GetDiscountedPrice(1000m)} руб.");
flexible.Transfer();
Console.WriteLine();

Ticket.DisplayTotalTickets();
```
